```python
def median_in_list(list1:list)->int:
    """in this queston we are going to find the median of the number"""
    if not isinstance(list1,list):
        raise TypeError("Input must be list")
    if len(list1)==0:
        raise ValueError("list cannot be empty")
    for i in list1:
        if not isinstance(i,(int,float)):
            raise ValueError("all element of list must be float or int")
    list1.sort()
    len_list=len(list1)
    mid=len_list//2
    if len_list %2!=0:
        return list1[mid]

    else:
        return (list1[mid-1]+list1[mid])/2
if __name__=="__main__":
    try:
        list1=list(map(int,input("enter the number in the list separated by space").split()))
        print(median_in_list(list1))
    except (TypeError,ValueError) as error:
        print(error)
    except Exception as e:
        print(e)
```
