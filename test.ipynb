```python
def second_largest(list1:list):
    """in this code we are going to find the second largest number from the list """
    largest=0
    second=0
    for i in list1:
        if i>largest:
            second=largest
            largest=i
        elif i >second and i<largest:
            second=i
    return second
if __name__=="__main__":
    try:
        list1=list(map(int,input("enter the number separatted by space").split()))
        print(second_largest(list1))
    except exception as e:
        print(e)
```

```python
def prime(num):
    """in this code we are going to check wheater the number is prime or not"""
    for i in range(2,num):
        if num%i==0:
            return "not prime "
            break
    else:
        return "prime"
if __name__=="__main__":
    try:
        num=int(input("enter the number"))
        print(prime(num))
    except Exception as e:
        print(e)
```

```python
# Find the first non repeated character in the string
def non_repeated_char(string):
    """in this code we are going to find the first non reapeated char in the string"""
    char_count={}
    for char in string:
        if char in char_count:
            char_count[char]+=1
        else:
            char_count[char]=1
    #return char_count
    for char in string:
        if char_count[char]==1:
            return char
if __name__=="__main__":
    try:
        string =input("enter the string").lower().replace(" ","")
        print(non_repeated_char(string))
    except Exception as e:
        print(e)
```
