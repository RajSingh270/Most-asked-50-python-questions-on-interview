# Remove NTH Occurrence of a character from a string


```python
#text="this is python class"
def remove_nth_occurrence(string: str,char_to_remove:str,n:int) -> str:
    occurrence_count=0
    result_characters=[]
    for current_character in string:
        if current_character==char_to_remove:
            occurrence_count += 1
            if occurrence_count==n: # 2nd positon
                continue
        result_characters.append(current_character)
    return "".join(result_characters)

if __name__=="__main__":
    try:
        string1=input("enter the text")
        print(remove_nth_occurrence(string1,"y",2))
    except Exception as e:
        print(e)
```

```python
def nth_occurence(string:str,char_to_remove:str,n:int)->str:
    """in this video we are going to find the nth occurence of a string and remove it"""
    count=0
    result=[]
    for i in string:
        if i==char_to_remove:
            count+=1
            if count==n:
                continue
        result.append(i)

    return "".join(result)
if __name__=="__main__":
    try:
        string=input("enter the string")
        print(nth_occurence(string,"y",2))
    except Exception as e:
        print(e)
```
