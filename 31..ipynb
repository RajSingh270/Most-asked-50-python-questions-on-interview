# Find the Sum of Squares of N Natural Numbers

```python
n=3
total=0
for i in range(0,n+1):
    total+=i**2
    i+=1
print(total)
```

```python
n=int(input("enter the number"))
print(n*(n+1)*(2*n+1)//6)
```

```python
def Square(num:int)->int:
    """in this code we are going to find the sum of 
    N natural numbers"""
    if not isinstance(num,int):
        raise TypeError ("number must be integer")
    if num <1:
        raise ValueError("num must be a positive number")
    if num ==0:
        raise ValueError("num is natural number so it is not start with zero")
    total=num*(num+1)*(2*num+1)//6 
    return total

if __name__=="__main__":
    try:
        
        num=int(input("enter the number"))
        print("Sum of Squares of N Natural Numbers",Square(num))

    except (TypeError,ValueError) as error:
        print(error)
    except Exception as e:
        print(e)
    
```
