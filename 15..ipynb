# Python Program to Check Leap Year

```python
def leap_year(year:int)->bool:
    """in this code we are going to cheak wheater the year is leap year or not"""
    if not isinstance(year,int):
        raise TypeError("Year must a integer value")
    if year <=1582:
        raise ValueError("year must be later of 1582")
    if (year%4==0 and year%100!=0)or year %400==0:
        return True
    else:
        return False
if __name__=="__main__":
    user_input=int(input("enter the year"))
    try:
        if leap_year(user_input):
            print(f"{user_input} is a leap year")
        else:
            print(f"{user_input} is not a leap year")
    except(TypeError,ValueError)as e:
        print(e)
    except Exception as e:
        print(e)
```
