```python
def find_the_sumdigit(n:int)->int:
    """in this code we are going to find the sum of the digit of the number"""
    if not isinstance(n,int):
        raise TypeError ("number enter must be in integer")
    if n<0:
        raise ValueError("number enter must be greater than zero")
    total=0
    while n>0:
        total=total+n%10
        n=n//10
    return total
if __name__=="__main__":
    n=input("enter the number")
    try: 
        n=int(n)
    except ValueError as error:
        print("error,please enter valid integer",error)
    else:
        try:
            print("sum of digit of the number is",find_the_sumdigit(n))
        except (TypeError,ValueError) as error:
            print(error)
    #except Exception as e:
        #print(e)
```
