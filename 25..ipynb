# Calculate Sum of a List of numbers using Recursion

```python
list1=[1,2,3,4,5]
sumlist=0
for i in list1:
    sumlist=sumlist+i
print(sumlist)
```

```python
# using  recursion
list1=[10,20,30,40]
def recursion(list1):
    if len(list1)==0:
        return 0
    return list1[0]+recursion(list1[1:])
```

```python
recursion(list1)
```
