# Swap two variables in python

```python
def swap_variable(a,b):
    a,b=b,a
    return a,b
if __name__=="__main__":
    try:
        a=int(input("enter the first variable a"))
        b=int(input("enter the second variable b"))
        print(swap_variable(a,b))
    except Exception as e:
        print(e)
```
