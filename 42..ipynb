```python
# num1=41234123
# num2=23421342343

def lcm_finder(num1:int,num2:int)->int:
    """in this code we are going to find the lcm of two number"""
    if not isinstance(num1,int)or not isinstance(num2,int):
        raise TypeError ("both numbers should be integer")
    if num1==0 or num2==0:
        raise ValueError("LCM is not definted for zeroes")
    def gcd(num1,num2):
        while num2!=0:
            num1,num2=num2,num1%num2
        return num1
    return abs(num1*num2)//gcd(num1,num2)

if __name__=="__main__":
    try:
        num1=int(input("enter the number1"))
        num2=int(input("enter the number2"))
        print(lcm_finder(num1,num2))
    except (TypeError,ValueError)as error:
        print(error)
    except Exception as e:
        print(e)
```
