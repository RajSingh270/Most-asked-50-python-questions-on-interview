```python
def reverse_an_integer(num:int)->int:
    """in this code we are going to find the reverse of an number or an integer"""
    if not isinstance(num,int):
        raise TypeError("input must be in integer")
    reverse=0
    sign=-1 if num<0 else 1
    num=abs(num)
    while num >0:
        digit=num%10
        reverse=reverse*10+digit
        num=num//10
    return reverse*sign

if __name__=="__main__":
    try:
        num=int(input("enter the number"))
        print("reverse of an integeris",reverse_an_integer(num))
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)
```
