# find the largest number from the list using for loop

```python
number_list=[1,22,332,1,1,2,3,3,42434,5235,23425,879,9879,1297319,1,13,13,131,313]
largest=number_list[0]
for i in number_list:
    if i>largest:
        largest=i
print(largest)
```

```python
def largest_num(number_list:list[float])->float:
    """function to find largest number in the list
    
    Args:
         number_list(number_list:list[float])a list containing numeric value (int or float)
         
    returns:
        Float:the largest number in the list
        
    Raises:
        TypeError:If input is not a list or contains non-numeric values
        ValueError:if the list is empty
        """
    #validate that input is a list
    if not isinstance(number_list,list):
        raise TypeError("Input must be a list of number float or int")

    # validate that list is not empty
    if not number_list:
        raise ValueError("LIst can not be empty")

    #validate that list elements should be numeric
    for num in number_list:
        if not isinstance(num,(int,float)):
            raise TypeError("all element must be numeric")
    #main logic
    largest=number_list[0]
    for num in number_list:
        if num>largest:
            largest=num
    return largest
# main execution
if __name__=="__main__":
    try:
        #taking input from the user
        user_input=input("enter numbers in the list seperate by space int or float only:")
        #convert to list or float
        number_list=[float(item) for item in user_input.split()]

        # calling the function
        result=largest_num(number_list)
        #display the result
        print("largest number in the list is :",result)

    except (TypeError ,ValueError)as error:
        print(error)
    except Exception as e:
        print(e)

    
```
