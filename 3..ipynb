```python
def is_prime_number(number:int)->bool:
    """function to check wheater a number is prime or not
    args:
        number(int:the number to check)
    return:
        bool:True if the number is prime,False otherwise.
    Raises:
        TypeError:If input is not an integer.
        Value:If number is less than 2(since primes are>=2).
    
    """
    # validate input an integer
    if not isinstance(number,int):
        raise TypeError("enter the integer number")
    if number<2:
        raise ValueError ("enter number must not be less than 2")
    for i in range(2, int(number**0.5) + 1):
        if number%i==0:
            return False
            break
    else:
        return True
# main execustion of code
if __name__=="__main__":
    try:
        user_input=input("enter the number to  check its prime or not")
        # validate and convert input to integer
        if not user_input.strip().isdigit():
            raise TypeError("input must be a digit(no letters no strings not float)")
        number=int(user_input)
        # calling the function
        if is_prime_number(number):
            print(f"{number}is a prime number")
        else:
            print(f"{number}is not a prime number")
    except (TypeError,ValueError)as error:
        print(error)
    except Exception as e:
        print("something went wrong",e)
    
is_prime_number(number)
```
