# Find the first non -repeated character in a string

```python
text="swwissn"
char_count={}
for char in text:
    if char in char_count:
        char_count[char]+=1
    else:
        char_count[char]=1
print(char_count)
for char in text:
    if char_count[char]==1:
        print(char)
        break
```

```python
def first_non_repeated(text:str)->str:
    char_count={}
    for char in text:
        if char in char_count:
            char_count[char]+=1
        else:
            char_count[char]=1
    print(char_count)
    for char in text:
        if char_count[char]==1:
            return char
if __name__=="__main__":
    try:
        text=input("Enter the text here")
        print(first_non_repeated(text))
    except Exception as e:
        print(e)
```
