# count each occurrence of characters in a sentence using recursion

```python
input_string="This is python class t"
input_string=input_string.lower()
char_count={}
for char in input_string:
    if char in char_count:
         char_count[char]+=1
    else:
        char_count[char]=1
print(char_count)
```

```python
def count_occurence(string:str)->dict:
    """count each occurrence of characters"""
    string=string.lower()
    char_count={}
    for char in string:
        if char in char_count:
            char_count[char]+=1
        else:
            char_count[char]=1
    return char_count
if __name__=="__main__":
    try:
        string=input("enter the string")
        print(count_occurence(string))
    except Exception as e:
        print(e)
```
