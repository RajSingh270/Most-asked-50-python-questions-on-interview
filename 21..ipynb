# python program to find gcd using recursion

```python
def gcd(a,b):
    """in this code we are going to find the gcd of two numbers using recursion"""
    if b==0:
        return a
    return gcd(b,a%b)
if __name__=="__main__":
    try:
        a=int(input("enter the num1"))
        b=int(input("enter the num2"))
        print(gcd(a,b))
    except Exception as e:
        print(e)
```
