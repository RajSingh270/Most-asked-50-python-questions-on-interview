# remove duplicate eliments from the list

```python
list1=[1,2,3,4,4,4]
remove=[]
for i in list1:
    if i not in remove:
        remove.append(i)
print(remove)
```

```python
def remove_duplicate(list1:list):
    """in this code we are going to 
    remove the duplicate values from the list"""
    if not isinstance(list1,list):
        raise TypeError ("the element must be in list")
    #if list1<[2]:
        #raise TypeError("list must contaian more that one element")
    remove=[]
    for i in list1:
        if i not in remove:
            remove.append(i)
    print(remove)

if __name__=="__main__":
    try:
        list1=list(map(str,input("enter the element separated by comma").split()))
    except (ValueError,TypeError)as error:
        print(error)
    except Exception as e:
        print(e)
remove_duplicate(list1)
    
```
