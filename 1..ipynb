# reverse the string using the for loop

```python
input_string="python"
reversed_string=""
for i in input_string:
    reversed_string=i+reversed_string
print(reversed_string)
```

```python
def reverse_string(input_string:str)->str:
    """
    function to reverse a givern string using for loop.
    args:
          input_str(str):the string to be reversed.
    returns:
          str:the reversed version of input string
    Raises:
           TypeError : If input is not a string
    """
    #validate the input type
    if not isinstance(input_string,str):
        raise TypeError ("input must be string")
    reversed_string=""
    for i in input_string:
        reversed_string=i+reversed_string
    return reversed_string
# ---Main Execution---
if __name__=="__main__":
    try:
        # taking input from the user
        user_input=input("enter your string to reverse:")
        #calling the function
        result=reverse_string(user_input)
        # Display the reversed string 
        print(f"reversed string={result}")
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)
        
```
