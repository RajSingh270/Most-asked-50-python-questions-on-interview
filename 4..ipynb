# print the prime number between 100 to 200

```python
def is_prime_number(number:int)->bool:
    """function to check wheater a number is prime or not
    args:
        number(int:the number to check)
    return:
        bool:True if the number is prime,False otherwise.
    Raises:
        TypeError:If input is not an integer.
        Value:If number is less than 2(since primes are>=2).
    
    """
    # validate input an integer
    if not isinstance(number,int):
        raise TypeError("enter the integer number")
    if number<2:
        raise ValueError ("enter number must not be less than 2")
    for i in range(2, int(number**0.5) + 1):
        if number%i==0:
            return False
            break
    else:
        return True
if __name__=="__main__":
    try:
        prime_list=[]
        start=input("enter the starting number")
        end=input("enter the ending number")
        if not start.strip().isdigit():
            raise TypeError("Start must  ve a valid integer(no decimal or integer) ")
        if not end.strip().isdigit():
            raise TypeError("End must be a valid integer (no decimals or letters)")
        start=int(start)
        end=int(end)
        for i in range(start,end):
            if is_prime_number(i):
                #print(f"{i} is a prime number")
                prime_list.append(i)
        print(prime_list)
    except (TypeError,ValueError) as error:
        print(error)
    except Exception as e:
        print(e)
```
