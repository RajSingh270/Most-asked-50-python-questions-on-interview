```python
def factorial(num:int)->int:
    """today i this video we are goint to learn about  the factorial
    now bund marao
    """
    if not isinstance(num,int):
        raise TypeError("num must be in integer")

    if num<0:
        raise ValueError("enter the valid number")

    factorial=1
    for i in range(2,num+1):
        factorial*=i
    return factorial
if __name__=="__main__":
    try:
        num=int(input("enter the number"))
        print(f"factorial of number is {factorial(num)}")
    except (TypeError,ValueError) as error:
        print(error)
    except Exception as e :
        print(e)
        
factorial(num)
```
