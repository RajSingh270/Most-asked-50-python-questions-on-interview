```python
def anagram(string1:str,string2:str):
    """in this code we are going to check wheather two string is anagram of not"""
    if not isinstance(string1,str)or not isinstance(string2,str):
        raise ValueError ("enter string must be word ")
    char_count={}
    if len(string1)!=len(string2):
        return False
    for char in string1:
        if char in char_count:
            char_count[char]+=1
        else:
            char_count[char]=1
    #return char_count
    for char in string2:
        if char not in char_count:
            return False
        char_count[char]-=1
        if char_count[char]<0:
            return False
            break
    else:
        return True
if __name__=="__main__":
    try:
        string1=input("enter the string").lower().replace(" ","")
        string2=input("enter the string2").lower().replace(" ","")
        print(anagram(string1,string2))
    except Exception as e:
        print(e)
```
