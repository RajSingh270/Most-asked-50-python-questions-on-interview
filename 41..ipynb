```python
example=[('apple', 5), ('banana', 2), ('cherry', 8), ('date', 3)]
```

```python
sorted(example,key = lambda x:x[1])
```

```python
def sort_dictionary(data1:dict):
    """in this code we are going to sort the dictionary on the basis of value"""
    if not isinstance(data1,dict):
        raise TypeError ("data must be in dictionary")
    return sorted(data1.items(),key=lambda x : x[1])
if __name__=="__main__":
    try:
        data1={'apple':5,'banana':2,'cherry':8,'date':3}
        print(sort_dictionary(data1))
    except Exception as e:
        print(e)
```
