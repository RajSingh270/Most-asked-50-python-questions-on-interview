```python
sentence="python is fun"
words_list=[]
word=""
for char in sentence:
    if char!=" ":
        word=word+char
    else: 
        words_list.append(word)
        word=""
words_list.append(word)
print(words_list)
reverse_sentence=""
for i in range(len(words_list)-1,-1,-1):
    print(i,words_list[i])
    reverse_sentence=reverse_sentence+words_list[i]
    if i!=0:
        reverse_sentence=reverse_sentence+" "
print(reverse_sentence)
```
