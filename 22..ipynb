```python
def decimal_to_binary(decimal:int):
    """in this code we are going to convert the decimal to binary"""
    if decimal==0:
        return "0"
    binary=""
    while decimal>0:
        remainder=decimal%2
        binary=str(remainder)+binary
        decimal=decimal//2
    return binary
if __name__=="__main__":
    try:
        decimal=int(input("enter the decimal number"))
        print(decimal_to_binary(decimal))
    except Exception as e:
        print(e)
```
