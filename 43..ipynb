```python
def binary_to_decimal(binary_str):
    total=0
    power=0
    for digit in reversed(binary_str):
        if digit=="1":
            
            total=total + 2**power
        power+=1
        
    return total
if __name__=="__main__":
    try:
        binary_str=input("enter the binary string")
        print(binary_to_decimal(binary_str))
    except Exception as e:
        print(e)
```
