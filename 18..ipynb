```python
def binary_search(list1:list,target:int)->int:
    """in this code we are going to find the binary search implementation"""
    if not isinstance(list1,list):
        raise TypeError ("sorted_list must contain list")
    if not isinstance(target,int):
        raise TypeError("target must be in integer")
    low=0
    high=len(list1)-1
    while low<=high:
        mid=(low+high)//2
        if list1[mid]==target:
            return mid
            break
        elif target <list1[mid]:
            high=mid-1
        else:
            low=mid+1
    return -1
if __name__=="__main__":
    try:
        target=int(input("enter the number"))
        list1=list(map(int,input("enter the number in the list").split(",")))
        print(binary_search(list1,target))
    except Exception as e:
        print(e)
```
