```python
def bubble_sort(list1: list) -> list:
    """Performs an optimized Bubble Sort on a list of integers."""
    for item in list1:
        if not isinstance(item, int):
            raise TypeError("All elements of list should be integers.")

    n = len(list1)
    for i in range(n):
        swapped = False
        for j in range(n - i - 1):
            if list1[j] > list1[j + 1]:
                list1[j], list1[j + 1] = list1[j + 1], list1[j]
                swapped = True
        
        # If no two elements were swapped, array is already sorted
        if not swapped:
            break

    return list1


if __name__ == "__main__":
    try:
        # Moved inside try block to catch conversion errors cleanly
        user_input = input("Enter numbers separated by space: ")
        list1 = list(map(int, user_input.split()))
        
        print("Sorted List:", bubble_sort(list1))
    except TypeError as error:
        print("TypeError:", error)
    except ValueError:
        print("ValueError: Please enter valid integers only.")
    except Exception as e:
        print("Error:", e)
```
