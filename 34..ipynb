# Find Longest Word In a Sentence

```python
text="""Building a modern application requires patience.Is programming difficult? Many people find the 
process rewarding Success comes to those who practice daily"""
```

```python
import string
#string.punctuation
text=text.lower()
pattern=str.maketrans("","",string.punctuation)
clean_text=text.translate(pattern)
#print(clean_text)
words_list=clean_text.split()
#print(words_list)
word_count_dict={}
for word in words_list:
    if word not in word_count_dict:
        word_count_dict[word]=len(word)
#print(word_count_dict)
max_len=max(word_count_dict.values())
longest_word=[key for key,value in word_count_dict.items()if value==max_len]
longest_word
```

```python
def char_frequency(s):
    freq = {}
    for ch in s:
        if ch != " ":
            if ch in freq:
                freq[ch] = freq[ch] + 1
            else:
                freq[ch] = 1
    return freq

s = input()
print(char_frequency(s))
```
