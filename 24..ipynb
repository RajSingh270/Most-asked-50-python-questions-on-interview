# Remove all whitespace from a string

```python
text="   \n       This     \t      is       \r           python"# This is python
result=""
prev_space=False
text.strip()
for char in text.strip():
    #print(char)
    if not char.isspace():
        result=result+char
        prev_space=False
    else:
        if prev_space==False:
            result=result+" "
            prev_space=True
print(result)
```
