# Count vowels in a String

```python
string =input('enter the string')
count=0
for i in string:
    if i=="A" or i=="E" or i=="O" or i=="I" or i=="U" or i=="a" or i=="e" or i=="i" or i=="o" or i=="u":
        count=count+1
print("total vowels in the string is",count)
```

```python
def count_vowels(text:str)->int:
    """count the number of vowels in the givern string
    parameters:
        text(str):the input string
    returns:
        int:Number of tvowels in the string
    Raises:
        TypeError:it the input is not a string
    """
    # validate if the input string or not
    if not isinstance(text,str):
        raise TypeError("text must be in words or string")
    # Main logic
    vowels="aeiou"
    count=0
    for char in text.lower():
        if char in vowels:
            count=count+1
    return count
# Main Execution
if __name__=="__main__":
    try:
        user_input=input("enter the text")

        result=count_vowels(user_input)
        print(f"Total vowels:{result}")
    except TypeError as error:
        print(error)
    except Exception as e:
        print(e)
    
        
```
