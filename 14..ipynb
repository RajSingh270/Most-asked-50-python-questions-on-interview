```python
def fibonacci_series(num:int)->list:
    """in this code we are going to find out the fibonnaci series"""
    fibonacci=[]
    a,b=0,1
    for i in range(num):
        fibonacci.append(a)
        a,b=b,a+b
    return fibonacci
if __name__=="__main__":
    try:
        num=int(input("enter the number"))
        print(fibonacci_series(num))
    except Exception as e:
        print(e)
```

```python
def fibonacci_series2(num:int)->list:
    """in this code we are going to find out the fibonnaci series"""
    a,b=0,1
    for i in range(num):
        print(a,end=" ")
        a,b=b,a+b
    print()
if __name__=="__main__":
    try:
        num=int(input("enter the number"))
        print(fibonacci_series2(num))
    except Exception as e:
        print("error",e)
```

```python
def fibonacci_series2(num: int) -> None:
    """Prints the Fibonacci series up to `num` terms."""
    a, b = 0, 1
    for _ in range(num):
        print(a, end=" ")
        a, b = b, a + b
    print()  # Adds a newline at the end

if __name__ == "__main__":
    try:
        num = int(input("Enter the number: "))
        fibonacci_series2(num)  # Simply call the function
    except Exception as e:
        print("Error:", e)
```
