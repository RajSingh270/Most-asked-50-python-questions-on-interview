```python
def flatten_list(nested_list:list)->list:
    """in this code we are going to flatten a list"""
    result=[]
    temp_list=list(nested_list)

    while temp_list:
        item =temp_list.pop()
        if isinstance (item,list):
            temp_list.extend(item)
        else:
            result.append(item)
            print("result :",result)
    print("final result: ",result)
    return result[::-1]

if __name__=="__main__":
    try:
        list1=[1,[2,[3,4],5],6]
        print(flatten_list(list1))
    except Exception as e:
        print(e)
```

```python
def flatten(nested:list)->list:
    """in this code we are going to flatte a nested list"""
    r=[]
    temp=list(nested)
    while temp:
        i=temp.pop()
        if isinstance (i,list):
            temp.extend(i)
        else:
            r.append(i)
    return r[::-1]

if __name__=="__main__":
    try:
        lis=[5,6,7,[21,32,[241,142,142,[412,412,4112]]]]
        print(flatten(lis))
    except Exception as e:
        print(e)
```
