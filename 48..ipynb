# Check if string contain only digits without using built in method

```python
def contain_only_digit(text:str)->bool:
    for char in text:
        if char<"0" or char >"9":
            return False
    else:
        return True

if __name__=="__main__":
    try:
        text=input("enter the text")
        print(contain_only_digit(text))
    except Exception as e:
        print(e)
```
