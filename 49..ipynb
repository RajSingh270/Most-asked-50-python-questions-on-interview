# Check If the number is armstrong number without converting number to string

```python
def is_armstrong(number:int)->bool:
    """this function tells wether a number armstrong or not"""
    temp_copy=number
    digit_count=0
    # step 1:count of digits
    while temp_copy>0:
        digit_count=digit_count + 1
        temp_copy=temp_copy//10
    print(digit_count)
    # step 2: total of digits 
    total=0
    temp_copy=number
    while temp_copy >0:
        digit =temp_copy % 10
        total=total + digit** digit_count
        temp_copy=temp_copy//10
    return total==number

if __name__=="__main__":
    try:
        number=int(input("Enter the number"))
        print(is_armstrong(number))
    except Exception as e:
        print(e)
```
