```python
def find_common_element(list1: list, list2: list, list3: list) -> list:
    """Returns a list of common elements in all three lists without using sets"""

    if not isinstance(list1, list) or not isinstance(list2, list) or not isinstance(list3, list):
        raise TypeError("all elements must be list")

    common = []

    for i in list1:
        if i in list2 and i in list3 and i not in common:
            common.append(i)

    return common


def convert_value(value):
    try:
        return int(value)
    except ValueError:
        return value


if __name__ == "__main__":
    try:
        list1 = [convert_value(x) for x in input("Enter elements in list1 separated by space: ").split()]
        list2 = [convert_value(x) for x in input("Enter elements in list2 separated by space: ").split()]
        list3 = [convert_value(x) for x in input("Enter elements in list3 separated by space: ").split()]

        print(find_common_element(list1, list2, list3))

    except TypeError as error:
        print(error)

    except Exception as e:
        print(e)
```
