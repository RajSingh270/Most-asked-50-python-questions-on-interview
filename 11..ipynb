# calculate the factorial of a number using recursion

```python
def factorial(n):
    """in this code we are going to find the facorial of a number"""
    if n==0 or n==1:#n==0 because factorial of 0 is 1 
        return 1
    return n* factorial(n-1)# recursive case

if __name__=="__main__":
    try:
        n=int(input("enter the number"))
        print(factorial(n))
    except Exception as e:
        print(e)
```
