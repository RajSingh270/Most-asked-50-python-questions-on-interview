# Generate list of even_numbers

```python
n=20
even_list=[]
for i in range(1,n+1):
    if i%2==0:
        even_list.append(i)
print(even_list)
```

```python
num=int(input("enter the number"))
even_number_list=[i for i in range(2,num+1,2)]
print(even_number_list)
```

```python
def even_list(num1:int)->list:
    """in this code we are going to geneate the even number list"""
    even_number_list1=[i for i in range(2,num1+1,2)]
    return even_number_list1
if __name__=="__main__":
    try:
        num1=int(input("enter the number"))
        print(even_list(num1))
    except Exception as e:
        print(e)
```
